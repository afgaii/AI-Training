## 8. Sets `{ }`

Unordered, mutable, **no duplicates**, supports set algebra. Elements must be hashable (no lists/dicts inside).

````python
sets = {"apple", 23, 5.6, False, ("a", "b")}
print(sets)

a = {1, 2, 3}
b = {2, 3, 4}
print(a.union(b))          # {1, 2, 3, 4}
print(a.intersection(b))   # {2, 3}
print(a.difference(b))     # {1}
print(a.symmetric_difference(b))  # {1, 4}
print(a.issubset({1, 2, 3, 4}))   # True
````

**Rule of thumb (LEGB):** Python looks up names in this order — **L**ocal → **E**nclosing → **G**lobal → **B**uilt-in.

Exactly. LEGB is the standard rule Python uses when resolving a name:

- Local — inside the current function
- Enclosing — in an outer function (for nested functions)
- Global — at the module level
- Built-in — names Python provides, such as len, print, and str

````python
x = "global" # global variable

def first():
    x = "enclosing" # enclosing variable

    def second():
        x = "local" # local variable
        print(x)
        
    second()
    
    print(x)

first()
print(x)
````

````python
# A closure: inner function "remembers" variables from outer scope
def make_multiplier(n):
    def multiplier(x):
        return x * n
        
    return multiplier

double = make_multiplier(2)
print(double(5))  # 10
````

## 14. Generators & Decorators

### Generators
A function becomes a generator automatically if it uses `yield` instead of `return`. Generators produce values lazily — one at a time, saving memory (crucial for large datasets).

````python
# Regular function -- builds the whole list in memory
def create_cubes(n):
    return [x**3 for x in range(n)]

print(create_cubes(3))
````

````python
# Generator -- yields one value at a time, doesn't store the whole list
def create_cubes_gen(n):
    for x in range(n):
        yield x**3
````

````python
for cube in create_cubes_gen(3):
    print(cube)
````

````python
# Regular function -- builds the whole list in memory
def create_cubes(n):
    return [x**3 for x in range(n)]

# Generator -- yields one value at a time, doesn't store the whole list
def create_cubes_gen(n):
    for x in range(n):
        yield x**3

for cube in create_cubes_gen(10):  # memory-efficient even at scale (try 1_000_000)
    print(cube, end=" ")
````

### Decorators
A decorator in Python is a function that takes another function as input, adds extra behavior, and returns a new function without changing the original code.

````python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

# Calling the decorated function
say_hello()

# Equivalent to: say_hello = my_decorator(say_hello)
````

````python
# Polymorphism -- same method name, different behavior per class
class Human:
    def speak(self):
        return "Hi"

class Cat:
    def speak(self):
        return "Meow"

animal = Human()
cat = Cat()

for x in [animal, cat]:
    print(x.speak())   # same call, different result -> polymorphism
````

````python
# Method Overriding -- child redefines a parent's method
class Animal:
    def speak(self):
        return "..."

class Cat(Animal):
    def speak(self):        # overrides Animal.speak
        return "Meow"

cat = Cat()
print(cat.speak())
````

- **Encapsulation** is about protecting data inside a class.
- It means keeping data (properties) and methods together in a class, while controlling how the data can be accessed from outside the class.
- This prevents accidental changes to your data and hides the internal details of how your class works.

````python
# Encapsulation
class Account:
    def __init__(self, balance):
        self.__balance = balance   # double underscore -> "private" (name-mangled)

    def get_balance(self):
        return self.__balance

acc = Account(1000)
print(acc.get_balance())

# Abstraction -- expose *what* it does, hide *how*
# e.g. calling model.fit() hides the entire gradient descent implementation from the user
````

- **Protected member:** convention prefix `_variable` — accessible within class and subclasses.
- **Private member:** convention prefix `__variable` — not meant to be accessed outside the class (Python enforces this only loosely, via name mangling).

## 18. `@staticmethod`, `@classmethod`, `@property`

- In Python, @staticmethod, @classmethod, and @property are mainly used to control how methods behave when accessed through a class or object.
1. @staticmethod — when you don't need the object or class
    - Normally, an instance method gets self
    - When to use: "This function logically belongs to this class, but it doesn't need any information from the object or class."

````python
class MathUtils:
    @staticmethod
    def add(a, b):              # doesn't need self or cls -- just a utility function
        return a + b

print(MathUtils.add(3, 4))
````

````python
# 2. @classmethod — when you need the class
class Person:
    population = 0

    def __init__(self, name):
        self.name = name
        Person.population += 1

    @classmethod
    def get_population(cls):    # operates on the class itself, not an instance
        return cls.population

p1 = Person("Ali")
p2 = Person("Sara")
print(Person.get_population())
````

````python
# 3. @property — when you want a method to look like an attribute
class Circle:
    def __init__(self, radius):
        self._radius = radius

    @property
    def area(self):             # lets you call circle.area like an attribute, not a method
        return 3.14159 * self._radius ** 2

c = Circle(5)
print(c.area)   # no parentheses needed -- reads like a plain attribute
````

| Decorator | Use when... |
|---|---|
| `@staticmethod` | The method doesn't need access to instance or class data |
| `@classmethod` | The method needs access to the class itself (e.g., alternate constructors) |
| `@property` | You want a method to be accessed like an attribute |

- The **try** block lets you test a block of code for errors.
- The **except** block lets you handle the error.

````python
try:
    x1 = 5
    x2 = 0
    print(x1/x2)
except ZeroDivisionError as e:
    print("Error!!!", e)
````

````python
try:
    x = 10 / 0
except ZeroDivisionError as e:
    print("Something went wrong:", e)
else:
    print("Nothing went wrong")   # runs only if no exception occurred
finally:
    print("This always runs")     # cleanup code, e.g. closing a file/connection
````

````python
class NotEnoughMoney(Exception):
    pass

def withdraw(balance, amount):
    if amount > balance:
        raise NotEnoughMoney("You have less money!!!")

    return balance - amount
````

````python
try:
    a = withdraw(500, 800)
except NotEnoughMoney as e:
    print(e)
else:
    print(a)
````

````python
# Custom exceptions
class InsufficientFundsError(Exception):
    pass

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError("Not enough funds")
    return balance - amount

try:
    withdraw(100, 500)
except InsufficientFundsError as e:
    print("Caught:", e)
````

## 24. Intro to Testing with `pytest`
- pytest is a Python testing framework. It helps you check whether your code works correctly.

Think of it like:
`Write code → write tests → run tests → see if code passes or fails`

1. Install pytest

- In your terminal:
```pip install pytest```

2. Create a Python function

- Suppose you have a file called calculator.py:
```python
def add(a, b):
    return a + b
```
3. Create a test

- Create another file called: `test_calculator.py`

- Inside:
```python
from calculator import add

def test_add():
    assert add(2, 3) == 5
```
The important part is: `assert add(2, 3) == 5`

It means:

- "I expect add(2, 3) to equal 5."

- If it does → PASS ✅

- If it doesn't → FAIL ❌

4. Run the test

- In your terminal, go to the folder containing your files and run:
```
pytest
```
You might see:

- 1 passed

- 🎉 Your test passed!

5. What if the function is wrong?

Suppose you accidentally write:
```python
def add(a, b):
    return a - b
```
Your test still expects:
`assert add(2, 3) == 5`

But 2 - 3 is -1, so pytest will report a failure.

- This is useful because it tells you that something is wrong with your cod

````python
# In a real project this would be two files:
# file: calculations.py
def add(a, b):
    return a + b

# file: test_calculations.py
def test_add():
    assert add(2, 3) == 5
    assert add(-1, 1) == 0

test_add()
print("All tests passed!")
# In practice, run from terminal: pytest test_calculations.py
````
